```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
data = {
    'Date': [
        '2026-09-25', '2026-09-26', '2026-09-27',
        '2026-09-28', '2026-09-29', '2026-09-30',
        '2026-10-01', '2026-10-01', '2026-10-02',
        '2026-10-02', '2026-10-03', '2026-10-03'
    ],

    'Category': [
        'Food', 'Travel', 'Shopping',
        'Food', 'Bills', 'Entertainment',
        'Travel', 'Food', 'Shopping',
        'Bills', 'Food', 'Travel'
    ],

    'Description': [
        'Lunch', 'Auto', 'Clothes',
        'Dinner', 'Internet', 'Movie',
        'Bus', 'Lunch', 'Groceries',
        'Mobile Recharge', 'Coffee', 'Auto'
    ],

    'Amount': [
        180, 120, 1200,
        250, 600, 350,
        60, 200, 750,
        299, 150, 130
    ]
}

df = pd.DataFrame(data)

df
```

```python
df['Date'] = pd.to_datetime(df['Date'])
```

```python
df.info()
```

```python
total_expense = df['Amount'].sum()

print("Total Expense: ₹", total_expense)
```

```python
category_expense = df.groupby('Category')['Amount'].sum()

print(category_expense)
```

```python
df.groupby('Category')['Amount'].sum()
```

```python
highest_category = category_expense.idxmax()
highest_amount = category_expense.max()

print("Highest Spending Category:", highest_category)
print("Amount Spent: ₹", highest_amount)
```

```python
average_expense = df['Amount'].mean()

print("Average Expense: ₹", round(average_expense, 2))
```

```python
largest_expense = df.loc[df['Amount'].idxmax()]

print("Largest Expense:")
print(largest_expense)
```

```python
category_expense.plot(kind='bar')

plt.title('Spending by Category')
plt.xlabel('Category')
plt.ylabel('Amount Spent (₹)')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
category_expense.plot(
    kind='pie',
    autopct='%1.1f%%'
)

plt.title('Expense Distribution by Category')
plt.ylabel('')
plt.show()
```

```python
daily_expense = df.groupby('Date')['Amount'].sum()

daily_expense.plot(
    kind='line',
    marker='o'
)

plt.title('Daily Spending Trend')
plt.xlabel('Date')
plt.ylabel('Amount Spent (₹)')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
print("===== PERSONAL EXPENSE ANALYSIS =====")

print("\nTotal Expense: ₹", total_expense)

print(
    "Average Expense: ₹",
    round(average_expense, 2)
)

print(
    "Highest Spending Category:",
    highest_category
)

print(
    "Highest Category Spending: ₹",
    highest_amount
)

print(
    "Largest Individual Expense: ₹",
    df['Amount'].max()
)
```
